```python

import networkx as nx
import matplotlib.pyplot as plt

# 1. Create a social network graph
G = nx.Graph()

users = ["Alice", "Bob", "Charlie", "David",
         "Emma", "Frank", "Grace"]

G.add_nodes_from(users)

# 2. Add connections between users
connections = [
    ("Alice", "Bob"),
    ("Alice", "Charlie"),
    ("Bob", "Charlie"),
    ("Bob", "David"),
    ("Charlie", "Emma"),
    ("David", "Emma"),
    ("Emma", "Frank"),
    ("Frank", "Grace"),
    ("Emma", "Grace")
]

G.add_edges_from(connections)

# 3. Display network information
print("SOCIAL MEDIA GRAPH ANALYSIS")
print("Number of users:", G.number_of_nodes())
print("Number of connections:", G.number_of_edges())

# 4. Find the most connected user
degrees = dict(G.degree())
print("\nConnections of each user:")
for user, degree in degrees.items():
    print(user, ":", degree)

popular_user = max(degrees, key=degrees.get)
print("\nMost connected user:", popular_user)

# 5. Find mutual friends
user1 = "Alice"
user2 = "Bob"

mutual_friends = (
    set(G.neighbors(user1))
    & set(G.neighbors(user2))
)

print("\nMutual friends of", user1, "and", user2,
      ":", mutual_friends)

# 6. Visualize the social network
plt.figure(figsize=(9, 6))
pos = nx.spring_layout(G, seed=42)

nx.draw(
    G, pos,
    with_labels=True,
    node_color="skyblue",
    node_size=2200,
    font_size=10,
    font_weight="bold",
    edge_color="gray"
)

plt.title("Social Media Graph Analysis")
plt.show()
```
