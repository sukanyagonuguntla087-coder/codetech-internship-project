```python

# CodeTech Task 2: Big Data Security Framework

import hashlib
import re

# Sample user records (fictional data)
records = [
    {"id": 1, "name": "Alice", "email": "alice@example.com",
     "role": "admin"},
    {"id": 2, "name": "Bob", "email": "bob@example.com",
     "role": "user"},
    {"id": 3, "name": "Charlie", "email": "invalid-email",
     "role": "user"}
]

# 1. Validate email addresses
def validate_email(email):
    pattern = r"^[\w.-]+@[\w.-]+\.[A-Za-z]{2,}$"
    return bool(re.match(pattern, email))

# 2. Mask email addresses
def mask_email(email):
    if not validate_email(email):
        return "INVALID EMAIL"
    name, domain = email.split("@", 1)
    return name[0] + "***@" + domain

# 3. Role-based access control
def access_data(role):
    if role == "admin":
        return "Access granted: full data access"
    elif role == "user":
        return "Access granted: limited data access"
    else:
        return "Access denied"

# 4. Hash a password example
def hash_password(password):
    return hashlib.sha256(password.encode()).hexdigest()

# Run security checks
print("BIG DATA SECURITY FRAMEWORK")
print("-" * 35)

for record in records:
    print("\nUser:", record["name"])
    print("Email valid:", validate_email(record["email"]))
    print("Masked email:", mask_email(record["email"]))
    print("Access:", access_data(record["role"]))

print("\nExample password hash:")
print(hash_password("DemoPassword123"))
```
